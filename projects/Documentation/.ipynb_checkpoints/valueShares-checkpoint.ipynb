### NOTE ON valueShares.InitTree

*The class iterates through a nesting structure with mixed input/output trees. It populates the tree with prices/quantities for intermediate goods as well as $\mu$ parameters for all nests. Parameter shares are defined such that the price index on all intermediate nests will be 1 and the quantity will reflect values. The following is a short extra explanation of the three types of iterations it goes through:*

Three types of iterations:

**Standard input-tree:**
* Apply if we have data (qD values) on all branches in the given nest.
* Get price and quantity of the knot. If not available (the case for intermediate nests) --> set price to 1 and quantity to the value of the sum of branches.
    * If the knot is an output --> look in 'pS','qS'
    * If the knot is an intermediate --> look in 'pD','qD'.
* Compute $\mu$ shares from basic CES formula.
* Add $\mu$ and (potential) new information on knots from (pS, qS, pD, qD) to the existing symbols, and remove the nests that we have dealt with from the nesting structure we are iterating through.

**Standard output-tree:**
* Apply if we have data (qD/qS values) on all branches in the given nest.
* Get price and quantity of the knot. If not available --> set price to 1 and quantity to sum of branches (always look in qD, pD).
* Compute $\mu$ shares from basic CET formula.
* Add $\mu$ and (potential) new information on knots from (pD, qD) to existing symbols. Remove nests that we deal wit here.

**Non-standard nests:**
* Apply when no progress is made using standard iterations (?).
* Apply for nests where an output nest branches into intermediate goods that are also branches in input nests *and* when we have data on the "knot" but not all branches.
* Get potential prices and quantities for relevant branches. Compute CET shares for these directly (this follows "standard" CET tree approach).
* For branches that we have not yet determined:
    * Assume prices of all these intermediate goods = 1.
    * Scale share parameters of all the undetermined branches to ensure price index of knot holds:
$$\begin{align}
    \overline{\mu}_{nn} = \frac{p_{n}^{1-\eta_n}-\sum_{j} \mu_{j,nn} p_j^{1-\eta_n}}{\sum_{l} \mu_{l,nn}^0 }, \qquad \mu_{l,nn} = \overline{\mu}_{nn} \mu_{l,nn}^0,
\end{align}$$
      where $j$ sums over branches with *known* share parameters and *l* identifies branches that we do not yet have data on, and $\mu_{l,nn}^0$ are "initial guesses" of these $\mu$'s.
    * Given share parameters, elasticities, and known values of the quantity/price of the relevant knot --> compute $q_{n,nn}$ from std. CET function.
* Finally: Add $\mu$ to existing symbol. Add new information on pD, qD. Remove nests that are dealt with.
